```python
import matplotlib.pyplot as plt

# Monthly sales data for a marketing company (in thousands of dollars)
monthly_sales = [45, 52, 48, 60, 72, 68, 80, 75, 70, 65, 58, 50]

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

# Calculate average sales
avg_sales = sum(monthly_sales) / len(monthly_sales)

# Plot sales data (red line)
plt.plot(months, monthly_sales, marker="o", color="red", label="Monthly Sales")

plt.title("Marketing Company Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales ($ thousands)")

# Average line
plt.axhline(y=avg_sales, linestyle="--", color="orange", label="Average Sales")

plt.legend()
plt.show()

```
